```python
def play_game():
    import random
    min = 1
    max = 100
    count = 0
    target = random.randint(1,100)
    print(target)
    print("=====猜數字遊戲=====\n")
    while(True):
        count +=1
        keyin = int(input("猜數字範圍{0}~{1}: ".format(min, max)))
        if(keyin >=min and keyin <= max):
            if(keyin == target):
                print("賓果!猜對了, 答案是:", target)
                print("您猜了",count,"次\n")
                break
            elif (keyin > target):
                max = keyin
                print("再小一點")
            elif (keyin < target):
                min = keyin
                print("再小一點")
            print("您猜了",count,"次\n")
        else:
            print("請輸入提示範圍內的數字")
    print("遊戲結束")
```

```python
while(True):
    play_game()
    is_play = input("你還要玩嗎?(y,n)")
    if is_play.upper() == "N":
        break
print("遊戲結束")
```
